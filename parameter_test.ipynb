# Hyperparameter - Trajectory Relationship
This simulation is intended to demonstrate the impacts of environmental hyperparameters on the ergodic trajectories generated.

```julia
using Simulator, Plots, Revise
```

## Spatial-only Matern-12 Kernel

```julia
# Define domain vars
dim = 100; # number of data points along each axis in the domain (defining a square domain)
xstart, xstop = 0, 25; # x-axis start and end positions
ystart, ystop = 0, 10; # y-axis start and end positions

σ_sq = 2.5; # variance
λₓ = 5.2; # spatial length scale

env_data = Simulator.SyntheticData.matern12_spatial(dim, xstart, xstop, ystart, ystop, σ_sq, λₓ);
heatmap(range(xstart, stop=xstop, length=dim), range(ystart, stop=ystop, length=dim), env_data)
title!("Synthetic Data: σ² = $(σ_sq), λₓ = $(λₓ)")
```

## Spatiotemporal Matern-12 Kernel

```julia
# Define domain vars
sdim = 10; # number of data points along each axis in the domain (defining a square domain)
tdim = 100; # number of temporal data points
xstart, xstop = 0, 25; # x-axis start and end positions
ystart, ystop = 0, 10; # y-axis start and end positions
tstart, tstop = 0, 3; # start and end times

σ_sq = 2.5; # variance
λₓ = 5.2; # spatial length scale
λₜ = 1.25; # temporal length scale

env_data = Simulator.SyntheticData.matern12_spatiotemporal(sdim, tdim, xstart, xstop, ystart, ystop, tstart, tstop, σ_sq, λₓ, λₜ);
```

```julia
minsol = minimum(env_data);
maxsol = maximum(env_data);
temporal_points = range(tstart, stop=tstop, length=tdim);
st_map = @animate for idx in 1:tdim
    heatmap(env_data[:,:,idx], clim=(minsol, maxsol));
    title!("Wind @ Time $(temporal_points[idx])");
end
gif(st_map)
```

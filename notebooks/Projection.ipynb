```julia
using LinearAlgebra
using Plots

# Define the point and line segment
point = [3, 2]
line_start = [1, 1]
line_end = [5, 1]

# Direction vector of the line segment
line_vec = line_end - line_start

# Vector from line_start to the point
point_vec = point - line_start

# Projection of point_vec onto line_vec
line_len_squared = dot(line_vec, line_vec)
projection_length = dot(point_vec, line_vec) / line_len_squared
projection_point = line_start + projection_length * line_vec

# Create the plot
plot([line_start[1], line_end[1]], [line_start[2], line_end[2]], seriestype=:line, label="Line Segment", color=:blue, lw=2)
scatter!([point[1]], [point[2]], color=:red, label="Point", marker=:circle, markersize=6)
scatter!([projection_point[1]], [projection_point[2]], color=:green, label="Projection Point", marker=:circle, markersize=6)

# Plot vectors using line segments
plot!([line_start[1], point[1]], [line_start[2], point[2]], label="Vector p", color=:black, linestyle=:dash)
plot!([line_start[1], line_end[1]], [line_start[2], line_end[2]], label="Direction Vector d", color=:blue, linestyle=:dot)
plot!([line_start[1], projection_point[1]], [line_start[2], projection_point[2]], label="Projection", color=:green, linestyle=:dot)

# Annotate plot
annotate!([(point[1], point[2]), (projection_point[1], projection_point[2])],
    ["Point", "Projection"],
    [(0, 10), (0, 10)],
    [(:red, 10), (:green, 10)])

xlabel!("x")
ylabel!("y")
title!("Vector Projection Visualization")
# legend(:topright)

# Display the plot
plot!()
```

# IIR Filter Testing
An IIR filter is of the form
$$
y_i = \alpha x_i + (1 - \alpha) y_{i-1}
$$
where $x_i$ is the current sensor reading, $y_{i-1}$ is the filterend measurement from the previous timestep, $0 \leq \alpha \leq 1$ is a filter coefficient, and $y_i$ is new filtered measurement.

```julia
α = 0.2;
x = 0.0;
y = 0.0;
xs = [0.0];
ys = [0.0];

Random.seed!(457)

for time in 1:40
    x = randn() * 5;
    push!(xs, x);

    y = α*x + (1 - α)*y;
    push!(ys, y);
end
```

```julia
plot(xs, label="Raw Sensor Data")
plot!(ys, label="Filtered Sensor Data")
title!("IIR Filter - α = $(α)")
```

```julia
iir_filter = @animate for α in 0:0.01:1
    x = 0.0;
    y = 0.0;
    xs = [0.0];
    ys = [0.0];

    Random.seed!(457)

    for time in 1:40
        x = randn() * 5;
        push!(xs, x);

        y = α*x + (1 - α)*y;
        push!(ys, y);
    end

    plot(xs, label="Raw Sensor Data")
    plot!(ys, label="Filtered Sensor Data")
    title!("IIR Filter - α = $(α)")
end
gif(iir_filter)
```
